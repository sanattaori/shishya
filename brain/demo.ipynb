# Teach Shishya something

A neural network that keeps learning after training. It reads a sentence once and knows it. No training runs.

Run the cells in order. A free CPU runtime is enough. The first cell downloads a free 400 MB model. Answers are slow, a few seconds per letter.


```python
!git clone --depth 1 https://github.com/sanattaori/shishya
%cd shishya
!pip install -q torch numpy huggingface_hub
from brain.teach import load, teach, answer
st = load()
```

## Teach it

Use a rare name. It stores only sentences with a word that is new to it, so a common name like Marie stores nothing.


```python
print(teach(st, "Priya lives in a town called Nashik."))
print(teach(st, "Zubak has a dog named Rex."))
```

## Ask it

Leave a sentence unfinished. Only that sentence goes in, so the answer can only come from its memory. Try your own wording.


```python
print(answer(st, "Priya lives in a town called"))
print(answer(st, "The name of Zubak's dog is"))
```

## Without its memory

The same network, not using what it read.


```python
print(answer(st, "Priya lives in a town called", memory=False))
```
